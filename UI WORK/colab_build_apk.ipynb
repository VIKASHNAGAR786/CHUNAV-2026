यह नोटबुक आपकी **100% शुद्ध हिंदी (बिना टूटी मात्राओं वाली)** और सम्पूर्ण 3,568 वोटर्स से युक्त Android APK को **मात्र 1 से 2 मिनट में तैयार करती है**।

--- 
### 🚀 APK बनाने के चरण:
1. **Cell 1** चलाएं: Android SDK, Java 17 व Gradle 8.4 सेटअप (~45 सेकंड)
2. **Cell 2** चलाएं: `UI_WORK_FOR_COLAB.zip` अपलोड करें
3. **Cell 3** चलाएं: APK कंपाइल करें (~45 सेकंड)
4. **Cell 4** चलाएं: आपकी तैयार `JAINAGAR_CHUNAV_2026.apk` तुरंत डाउनलोड हो जाएगी!

```python
# Step 1: Install Java 17, Android SDK & Official Gradle 8.4
!apt-get update -qq
!apt-get install -y -qq openjdk-17-jdk wget unzip

import os
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-amd64"
os.environ["ANDROID_HOME"] = "/usr/local/android-sdk"

# 1. Download official Gradle 8.4 (takes ~3 seconds on Colab)
if not os.path.exists("/opt/gradle-8.4/bin/gradle"):
    print("⚡ Downloading official Gradle 8.4...")
    !wget -q https://services.gradle.org/distributions/gradle-8.4-bin.zip -O /tmp/gradle-8.4.zip
    !unzip -q /tmp/gradle-8.4.zip -d /opt/
    !rm -f /tmp/gradle-8.4.zip

os.environ["PATH"] = f"/opt/gradle-8.4/bin:/usr/local/android-sdk/cmdline-tools/latest/bin:/usr/local/android-sdk/platform-tools:{os.environ['JAVA_HOME']}/bin:" + os.environ["PATH"]

# 2. Download Android Command Line Tools
if not os.path.exists("/usr/local/android-sdk/cmdline-tools/latest/bin/sdkmanager"):
    print("⚡ Setting up Android SDK Tools...")
    !mkdir -p /usr/local/android-sdk/cmdline-tools
    !wget -q https://dl.google.com/android/repository/commandlinetools-linux-11076708_latest.zip -O /tmp/cmdline.zip
    !unzip -q /tmp/cmdline.zip -d /usr/local/android-sdk/cmdline-tools/
    !mv /usr/local/android-sdk/cmdline-tools/cmdline-tools /usr/local/android-sdk/cmdline-tools/latest 2>/dev/null || true
    !rm -f /tmp/cmdline.zip
    !yes | sdkmanager --licenses > /dev/null
    !sdkmanager "platforms;android-34" "build-tools;34.0.0" > /dev/null

!gradle -v
print("\n✅ [Step 1 Completed] Java 17, Android SDK & Gradle 8.4 are ready!")
```

```python
# Step 2: Upload UI_WORK_FOR_COLAB.zip
import os, zipfile
from google.colab import files

if not os.path.exists("android_project"):
    print("👉 'Choose Files' पर क्लिक करके 'UI_WORK_FOR_COLAB.zip' अपलोड करें:")
    uploaded = files.upload()
    for fname in uploaded.keys():
        if fname.endswith(".zip"):
            with zipfile.ZipFile(fname, 'r') as zip_ref:
                zip_ref.extractall(".")
            print(f"✅ Extracted: {fname}")
else:
    print("✅ android_project already exists in working directory!")

assert os.path.exists("android_project/app/build.gradle"), "Error: android_project not found!"
print("\n✅ [Step 2 Completed] All project files verified successfully!")
```

```python
# Step 3: Fast Compile APK via Gradle 8.4 (~45 seconds)
import os

# Self-heal check: ensure Gradle 8.4 is downloaded even if Step 1 was skipped
if not os.path.exists("/opt/gradle-8.4/bin/gradle"):
    print("⚡ Downloading official Gradle 8.4...")
    !wget -q https://services.gradle.org/distributions/gradle-8.4-bin.zip -O /tmp/gradle-8.4.zip
    !unzip -q /tmp/gradle-8.4.zip -d /opt/
    !rm -f /tmp/gradle-8.4.zip

os.environ["PATH"] = f"/opt/gradle-8.4/bin:/usr/local/android-sdk/cmdline-tools/latest/bin:/usr/local/android-sdk/platform-tools:/usr/lib/jvm/java-17-openjdk-amd64/bin:" + os.environ["PATH"]
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-amd64"
os.environ["ANDROID_HOME"] = "/usr/local/android-sdk"

%cd /content/android_project
!/opt/gradle-8.4/bin/gradle assembleDebug --no-daemon

import glob
apks = glob.glob("/content/android_project/app/build/outputs/apk/debug/*.apk")
if apks:
    print(f"\n🎉 [Step 3 Completed] SUCCESS! APK created in under 1 minute!")
    print(f"Generated APK: {apks[0]}")
else:
    print("❌ Check logs above for details.")
```

```python
# Step 4: Download compiled APK to your PC / Phone
import glob, shutil
from google.colab import files

apks = glob.glob("/content/android_project/app/build/outputs/apk/debug/*.apk")
if apks:
    dest_apk = "/content/JAINAGAR_CHUNAV_2026.apk"
    shutil.copy2(apks[0], dest_apk)
    print(f"⬇️ Downloading {dest_apk} to your browser...")
    files.download(dest_apk)
    print("✅ Done! Transfer this APK to your Android phone and install it.")
else:
    print("❌ APK file not found to download.")
```
