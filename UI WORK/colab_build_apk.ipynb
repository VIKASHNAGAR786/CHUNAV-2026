This Google Colab notebook compiles your complete offline Kivy Android application (`JAINAGAR_CHUNAV_2026.apk`) using **Buildozer** in the cloud.

--- 
### 🚀 Steps to build your APK:
1. Run **Cell 1**: Install System Buildozer & Android Dependencies (takes ~2 mins).
2. Run **Cell 2**: Upload your `UI_WORK.zip` or files.
3. Run **Cell 3**: Execute Buildozer Android Debug Build (takes ~10 mins on first build).
4. Run **Cell 4**: Download your `JAINAGAR_CHUNAV.apk` directly to your phone or PC!

```python
# Step 1: Install System & Buildozer Dependencies
!sudo apt-get update -y
!sudo apt-get install -y \
    git \
    zip \
    unzip \
    openjdk-17-jdk \
    python3-pip \
    autoconf \
    libtool \
    pkg-config \
    zlib1g-dev \
    libncurses5-dev \
    libncursesw5-dev \
    libtinfo5 \
    cmake \
    libffi-dev \
    libssl-dev

!pip install --upgrade Cython==0.29.36 buildozer
print("\n System dependencies & Buildozer installed successfully!")
```

```python
# Step 2: Upload or prepare project directory
import os
from google.colab import files

# Check if project files are already present
if not os.path.exists("main.py"):
    print("Please upload your 'UI_WORK.zip' archive:")
    uploaded = files.upload()
    for fname in uploaded.keys():
        if fname.endswith(".zip"):
            !unzip -q "{fname}"
            print(f"Extracted {fname}!")
else:
    print("Project files already present in working directory!")

# Verify critical files
assert os.path.exists("main.py"), "main.py not found!"
assert os.path.exists("buildozer.spec"), "buildozer.spec not found!"
assert os.path.exists("jainagar_chunav.db"), "jainagar_chunav.db not found!"
print(" All critical project files verified and ready for build!")
```

```python
# Step 3: Run Buildozer to compile the Android APK
!buildozer -v android debug
```

```python
# Step 4: Locate & Download compiled APK to your device
import glob
import os
from google.colab import files

apks = glob.glob("bin/*.apk")
if apks:
    target_apk = apks[0]
    print(f" APK Found: {target_apk} ({round(os.path.getsize(target_apk)/(1024*1024), 2)} MB)")
    print("Starting download to your browser...")
    files.download(target_apk)
else:
    print(" No APK found in 'bin/' directory. Please check Step 3 build logs for details.")
```
